```python
# 한 문자만 인덱싱 : 변수명[인덱스번호]
a = 'Hello Python'
print(a[0])
print(a[6])
print(a[-1])
```

```python
# 연속된 부분의 문자열을 슬라이싱 : 변수명[시작인덱싱:종료인덱스]
a = 'Hello Python'
print(a[0:5])          # 0 ~ 4
print(a[6:])           # 6 ~ 끝
print(a[:5])           # 0 ~ 4
print(a[6:-4])         # 6 ~ 끝에서 5번째
print(a[0:9:2])        # 0 ~ 8(2간격)
```
